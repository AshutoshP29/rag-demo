# DOCUMENT STRUCTURE
A type of data structure that contains two important information, that is, Page Content(String) and Metadata(dictionary).
Embedding is performed on the page_content of each Document, and not on the entire Document object. The Document object (including its metadata) is then stored together with the resulting embedding in the vector database.

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()
print(document)
```

```python
from langchain_community.document_loaders import DirectoryLoader

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob = "**/*.txt", ## Pattern to match files, our filter
    loader_cls = TextLoader, ## Loader class to use
    loader_kwargs = {'encoding':'utf-8'},
    show_progress = False
)

documents = dir_loader.load()
documents
```
